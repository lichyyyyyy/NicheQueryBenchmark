```python
import scanpy as sc

adata_abca_1 = sc.read_h5ad('data/zhuang-abca-1/Zhuang-ABCA-1-raw.h5ad')
adata_abca_1
```

```python
import pandas as pd

cells_abca_1 = pd.read_csv('data/zhuang-abca-1/cell_metadata.csv')
cells_abca_1
```

```python
import pandas as pd

ccf_1 = pd.read_csv('data/zhuang-abca-1/ccf_coordinates.csv')
ccf_1
```

```python
# how many slices does zhuang-abca-1 contain? How many cells per sample?
# `abc_sample_id`: an individual sample (animal / batch / experiment)
# `brain_section_label`: a brain slice of one sample

print('How many slices does `zhuang-abca-1` contain? How many cells per slice?\n')
print(
    cells_abca_1["brain_section_label"].value_counts().describe(percentiles=[.01, .05, .1, .25, .5, .75, .9, .95, .99]))
```

```python
# how many brain sections does zhuang-abca-1 contain?


print('How many brain sections does `zhuang-abca-1` contain? How many cells per brain section?\n')
print(ccf_1["parcellation_index"].value_counts().describe(percentiles=[.01, .05, .1, .25, .5, .75, .9, .95, .99]))
```

```python
# given a brain section, how many samples include the section and how many cells are in this section per sample?
import pandas as pd

ccf_cell_1 = pd.merge(cells_abca_1, ccf_1, how='left', on='cell_label')
ccf_cell_1.groupby('parcellation_index')

print('Given a brain section, how many slices include the section?\n')
print(ccf_cell_1.groupby('parcellation_index')['brain_section_label'].nunique().describe(
    percentiles=[.01, .05, .1, .25, .5, .75, .9, .95, .99]))
```

```python
from NicheQueryPrototype.database import Database

parcellation_tree = Database().parse_parcellation_structure(json_path='data/parcellation/structure.json')
ccf_cell_1['parcellation_level'] = ccf_cell_1['parcellation_index'].map(
    lambda k: parcellation_tree.get(k, {}).get("st_level", -1))

s = (ccf_cell_1
     .groupby(["parcellation_level", "parcellation_index"])["brain_section_label"]
     .nunique())

print('Given a brain section, how many slices include the section group by brain section level?\n')
print(s.groupby(level=0).describe(percentiles=[.01, .05, .1, .25, .5, .75, .9, .95, .99]))
```

```python
s = (ccf_cell_1
     .groupby(["parcellation_level"])["parcellation_index"]
     .nunique())

print('# parcellations group by levels\n')
print(s)
```

```python
s = (ccf_cell_1
     .groupby(["parcellation_level", "parcellation_index"])["cell_label"]
     .nunique())

print('Given a brain section, how many cells in the brain section per brian slice group by brain section level?\n')
print(s.groupby(level=0).describe(percentiles=[.01, .05, .1, .25, .5, .75, .9, .95, .99]))
```
